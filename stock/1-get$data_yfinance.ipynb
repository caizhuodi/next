## 1.1 ^IXIC 纳指 NASDAQ Composite 

### 1.1.1 since a specific day
🌟 since a specific day, e.g. 1985-01-01

````python
# IMPORT THE LIBRARY
import yfinance as yf
from datetime import datetime

# CREATE TICKER INSTANCE FOR the stock you select
ticker = yf.Ticker("^IXIC") # iShares Silver Trust (SLV); NASDAQ Composite (^IXIC)

# GET TODAYS DATE AND CONVERT IT TO A STRING WITH YYYY-MM-DD FORMAT (YFINANCE EXPECTS THAT FORMAT)
end_date = datetime.now().strftime('%Y-%m-%d')
history_1d = ticker.history(start='1985-01-01',end=end_date)
print(history_1d)
````

````python
import pandas as pd

history_1d.to_csv(f'./data/^IXIC_history_1d_{end_date}.csv', index=True)
````

### 1.1.2 set interval

🌟 set interval, e.g. 1m

❗️1m data is only retrievable for the last 7 days

> High granularity of data. One cool feature of yfinance is that you can get highly refined data, all the way down to 5 minute, 3 minute and even 1 minute data! The full range of intervals available are: ```1m, 2m, 5m, 15m, 30m, 60m, 90m, 1h, 1d, 5d, 1wk, 1mo, 3mo```. However it is important to note that the 1m data is only retrievable for the last 7 days, and anything intraday (interval<1d) only for the last 60 days. - https://algotrading101.com/learn/yfinance-guide/

````python
# GETTING THE MAX AMOUNT OF RECENT 1 MINUTE DATA.
# OBSERVATION: THE VARIABLE 'end_date' WAS DECLARED IN THE PREVIOUS CODE SNIPPED
history_1m = ticker.history(period='max',end=end_date,interval='1m')
print(history_1m)
````

````python
import pandas as pd

history_1m.to_csv(f'./data/^VIX_history_1m_{end_date}.csv', index=True)
````

## 1.2 ^VIX 恐慌指数
🌟 earliest date 1900-01-01

````python
# IMPORT THE LIBRARY
import yfinance as yf
from datetime import datetime

from curl_cffi import requests
session = requests.Session(impersonate="chrome")

# CREATE TICKER INSTANCE FOR the stock you select
ticker = yf.Ticker("^VIX", session=session)

# GET TODAYS DATE AND CONVERT IT TO A STRING WITH YYYY-MM-DD FORMAT (YFINANCE EXPECTS THAT FORMAT)
end_date = datetime.now().strftime('%Y-%m-%d')
history_1d = ticker.history(start='1900-01-01',end=end_date)
print(history_1d)

history_1m = ticker.history(period='max',end=end_date,interval='1m')
print(history_1m)
````

````python
import pandas as pd

history_1d.to_csv(f'./data/^VIX_history_1d_{end_date}.csv', index=True)
history_1m.to_csv(f'./data/^VIX_history_1m_{end_date}.csv', index=True)
````
